```python
def showBoard(board):
    print()
    print(f" {board[0] or ' '} | {board[1] or ' '} | {board[2] or ' '} ")
    print("---+---+---")
    print(f" {board[3] or ' '} | {board[4] or ' '} | {board[5] or ' '} ")
    print("---+---+---")
    print(f" {board[6] or ' '} | {board[7] or ' '} | {board[8] or ' '} ")
    print()

showBoard(board)
```

```python
winningLines = [
    [0,1,2],
    [3,4,5],
    [6,7,8],
    [0,3,6],
    [1,4,7],
    [2,5,8],
    [0,4,8],
    [2,4,6]
]

def winnerBoard(board):
    for a, b, c in winningLines:
        if board[a] != '' and board[a] == board[b] == board[c]:
            return board[a]
    return None

def emptyBoxes(board):
    return [i for i in range(9) if board[i] == '']
```

```python
def minimax(board, is_max_turn, alpha, beta):
    win = winnerBoard(board)

    if win == 'X':
        return 1

    if win == 'O':
        return -1

    if len(emptyBoxes(board)) == 0:
        return 0

    if is_max_turn:
        best_score = -2

        for box in emptyBoxes(board):
            board[box] = 'X'

            score = minimax(board, False, alpha, beta)

            board[box] = ''

            best_score = max(best_score, score)
            alpha = max(alpha, best_score)

            if beta <= alpha:
                break

        return best_score

    else:
        best_score = 2

        for box in emptyBoxes(board):
            board[box] = 'O'

            score = minimax(board, True, alpha, beta)

            board[box] = ''

            best_score = min(best_score, score)
            beta = min(beta, best_score)

            if beta <= alpha:
                break

        return best_score
```

```python
def best_move(board):
    best_score = -2
    best_box = None

    for box in emptyBoxes(board):
        board[box] = 'X'

        score = minimax(board, False, -2, 2)

        board[box] = ''

        if score > best_score:
            best_score = score
            best_box = box

    return best_box


test_board = [
    'X', 'O', 'X',
    'O', 'X', '',
    '', '', 'O'
]

showBoard(test_board)

move = best_move(test_board)

print('AI chooses position:', move)
```

```python
winningLines = [
    [0,1,2],
    [3,4,5],
    [6,7,8],
    [0,3,6],
    [1,4,7],
    [2,5,8],
    [0,4,8],
    [2,4,6]
]


def winnerBoard(board):
    for a, b, c in winningLines:
        if board[a] != '' and board[a] == board[b] == board[c]:
            return board[a]
    return None


def emptyBoxes(board):
    return [i for i in range(9) if board[i] == '']


def minimax(board, is_max_turn, alpha, beta):
    win = winnerBoard(board)

    if win == 'X':
        return 1

    if win == 'O':
        return -1

    if len(emptyBoxes(board)) == 0:
        return 0

    if is_max_turn:
        best_score = -2

        for box in emptyBoxes(board):
            board[box] = 'X'

            score = minimax(board, False, alpha, beta)

            board[box] = ''

            best_score = max(best_score, score)
            alpha = max(alpha, best_score)

            if beta <= alpha:
                break

        return best_score

    else:
        best_score = 2

        for box in emptyBoxes(board):
            board[box] = 'O'

            score = minimax(board, True, alpha, beta)

            board[box] = ''

            best_score = min(best_score, score)
            beta = min(beta, best_score)

            if beta <= alpha:
                break

        return best_score


def best_move(board):
    best_score = -2
    best_box = None

    for box in emptyBoxes(board):
        board[box] = 'X'

        score = minimax(board, False, -2, 2)

        board[box] = ''

        if score > best_score:
            best_score = score
            best_box = box

    return best_box


def play_game():
    board = [''] * 9

    print('You are O. The AI is X.')
    print('Choose positions using numbers 0 to 8.')

    while True:

        # AI move
        ai_move = best_move(board)
        board[ai_move] = 'X'

        print('\nAI chooses:', ai_move)
        showBoard(board)

        if winnerBoard(board) == 'X':
            print('AI wins!')
            break

        if len(emptyBoxes(board)) == 0:
            print('Draw!')
            break

        # Human move
        while True:
            try:
                player_move = int(input('Your move (0-8): '))

                if player_move in emptyBoxes(board):
                    break

                print('That position is not available.')

            except ValueError:
                print('Please enter a number from 0 to 8.')

        board[player_move] = 'O'

        showBoard(board)

        if winnerBoard(board) == 'O':
            print('You win!')
            break

        if len(emptyBoxes(board)) == 0:
            print('Draw!')
            break


play_game()
```
