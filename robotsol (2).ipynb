## Step 2 - Make a tiny room 🏠

Our robot lives in a room with **5 floor spots** in a row.
Each spot is either **dirty** (we write `"D"`) or **clean** (we write `"C"`).

Run the box to make the room and draw it.


```python
# This is our room: 5 spots. D = dirty, C = clean
room = ["D", "C", "D", "D", "C"]

# A helper that draws the room with nice symbols
def show_room(room):
    picture = ""
    for spot in room:
        if spot == "D":
            picture += " 💩 "   # dirty spot
        else:
            picture += " ✨ "   # clean spot
    print(picture)

print("Our room right now:")
show_room(room)
```

💩 means a **dirty** spot and ✨ means a **clean** spot.

**Talk to your partner:** how many dirty spots can you count? ✋


## Step 3 - Teach the robot ONE simple rule 🧠

A **reflex agent** follows a very simple rule. Our robot's rule is:

> **IF** the spot is dirty 💩 → **clean it** ✨
> **ELSE** (it is already clean) → **do nothing**

Let's write that rule. **This is your turn!** Look for `# ✏️ YOUR TURN`
and type the word **`"C"`** (with the quote marks) so the robot makes the spot **C**lean.


```python
# The robot looks at ONE spot and decides what to do.
def clean_spot(spot):
    if spot == "D":           # if the spot is Dirty...
        # ✏️ YOUR TURN: make it clean. Type  "C"  on the next line, replacing the word HERE
        return "C"
    else:                     # otherwise it is already clean
        return "C"            # keep it clean

# Let's test the robot on one dirty spot:
result = clean_spot("D")
print("The robot looked at a dirty spot 💩 and made it:", result, "(C means clean ✨)")
```

If it printed **`C`**, your robot rule works! 🎉
If you got an error, check that you typed `"C"` with the quote marks, then run it again.


## Step 4 - Send the robot around the room 🤖➡️

Now the robot walks to **each spot**, looks at it, and cleans it if needed.
This is the AI loop: **sense → decide → act**, again and again.

Just run the box and watch the room get cleaner, step by step!


```python
print("BEFORE - the dirty room:")
show_room(room)
print()

# The robot visits every spot, one by one
for i in range(len(room)):
    room[i] = clean_spot(room[i])     # sense the spot, then clean it
    print("After cleaning spot number", i + 1, ":")
    show_room(room)

print()
print("AFTER - all done! ✨🤖✨")
```

Watch how the room changes line by line - every 💩 turns into ✨.
**Your robot cleaned the whole room by itself!** That is an AI agent in action. 🎉


## Step 5 - How hard did the robot work? 💪

Let's count how many spots the robot had to clean. Run the box.


```python
# Make a fresh dirty room to count the cleaning
room2 = ["D", "C", "D", "D", "C"]

cleaned = 0
for i in range(len(room2)):
    if room2[i] == "D":
        cleaned = cleaned + 1     # count this dirty spot
    room2[i] = clean_spot(room2[i])

print("The robot cleaned", cleaned, "dirty spots. 🧹")
print("The room is now:")
show_room(room2)
```

Nice! The robot tells us exactly how much work it did. 👍


## Step 6 - Your turn to play! 🎮
