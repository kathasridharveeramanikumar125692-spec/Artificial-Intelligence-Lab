```python
# c = Clean
# d = Dirty

room = ["c", "d", "d", "c", "d"]

def show_room(room):
    picture = ""
    for spot in room:
        if spot == "d":
            picture += "😱 "
        else:
            picture += "😎 "
    print(picture)

print("Our room right now:")
show_room(room)
```

```python
# Count of cleaned spots
count = 0

def clean_spot(spot):
    global count

    if spot == "d":
        count += 1
        return "c"
    else:
        return "c"

result = clean_spot("d")
print("The robot cleaned the dirty spot:", result)
print("Dirty spots cleaned:", count)
```

```python
# Reset room and counter
room = ["c", "d", "d", "c", "d"]
count = 0

print("BEFORE - Dirty Room:")
show_room(room)
print()

for i in range(len(room)):
    room[i] = clean_spot(room[i])
    print(f"After cleaning spot {i + 1}:")
    show_room(room)

print()
print("AFTER - All Done 😁")
show_room(room)

print("\nTotal dirty spots cleaned:", count)
```
