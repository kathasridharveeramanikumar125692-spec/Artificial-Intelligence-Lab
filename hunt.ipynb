```python
# The map: each room lists the rooms you can walk to next.
map_rooms = {
    "A": ["B", "C"],   # from A you can go to B or C
    "B": ["D"],        # from B you can go to D
    "C": ["D"],        # from C you can go to D
    "D": ["G"],        # from D you can go to G
    "G": [],           # G is the treasure! No need to go further.
}

print("Our map:")
print("            A  (Start)")
print("           / \\")
print("          B   C")
print("           \\ /")
print("            D")
print("            |")
print("            G  (Treasure 💎)")
print()
print("Doors from each room:", map_rooms)
```

```python
def bfs(start, goal):
    to_do = [start]
    visited = []
    order = []

    while to_do:
        room = to_do.pop(0)

        if room not in visited:
            visited.append(room)
            order.append(room)

            if room == goal:
                break

            for next_room in map_rooms[room]:
                if next_room not in visited and next_room not in to_do:
                    to_do.append(next_room)

    return order

order = bfs("A", "G")

print("BFS checked the rooms in this order:", order)
print("Number of rooms BFS checked:", len(order))
```

```python
def dfs(start, goal):
    to_do = [start]
    visited = []
    order = []

    while to_do:
        room = to_do.pop()

        if room not in visited:
            visited.append(room)
            order.append(room)

            if room == goal:
                break

            for next_room in map_rooms[room]:
                if next_room not in visited and next_room not in to_do:
                    to_do.append(next_room)

    return order

order = dfs("A", "G")

print("DFS checked the rooms in this order:", order)
print("Number of rooms DFS checked:", len(order))
```

```python
door_cost={
    ("A","B"):1,
    ("B","D"):1,
    ("A","C"):5,
    ("C","D"):1,
    ("D","G"):1,

}
```

```python

# Function to calculate path cost
def path_cost(path):
    total = 0
    for i in range(len(path) - 1):
        door = (path[i], path[i + 1])
        total += door_cost[door]
    return total

# Paths
path1 = ["A", "B", "D", "G"]
path2 = ["A", "C", "D", "G"]   

# Display results
print("Path 1 A -> B -> D -> G costs:", path_cost(path1))
print("Path 2 A -> C -> D -> G costs:", path_cost(path2))
```

```python
cost1 = path_cost(path1)
cost2 = path_cost(path2)

# ✏️ YOUR TURN: we want the cheaper path. Replace HERE with the < sign.
if cost1 < cost2:
    best_path = path1
    best_cost = cost1
else:
    best_path = path2
    best_cost = cost2

print("Path 1 cost:", cost1)
print("Path 2 cost:", cost2)
print()
print("The cheapest path is:", best_path, "with cost", best_cost, "💰")

bfs_order = bfs("A", "G")
dfs_order = dfs("A", "G")

print("RESULTS")
print("-" * 40)
print("BFS  (nearest first) checked:", len(bfs_order), "rooms ->", bfs_order)
print("DFS  (deep first)    checked:", len(dfs_order), "rooms ->", dfs_order)
print("Cheapest path:", best_path, "with cost", best_cost)
print("-" * 40)
print()
print("What we learned:")
print(" * BFS checks the nearest rooms first (spreads out).")
print(" * DFS dives deep down one path first.")
print(" * Cheapest path counts door costs and picks the lowest total.")
```
